```python
import pandas as pd
import matplotlib.pyplot as plt

# Load a dataset (for example, a CSV file)
df = pd.read_csv('mydata.csv')

# Quick look at the data
display(df.head())         # shows first 5 rows of the data
display(df.describe())     # summary statistics of numerical columns

# Plot a histogram of a column (e.g., "age")
plt.hist(df['age'], bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```
